***Question***: Use numpy to build an automated real estate appraisal engine instead of relying on slow, manual property comparisons, this project uses multidementional NumPy arrays to ingest raw houusing data (square footage, bedrooms, age, proximity to city centre), normalize the scales and calculate a ' similarity index' to instantly value an unpriced home based on its nearest neibour


```python
# import numpy
import numpy as np
#My defined house data
# The features per row are; sqr feet, bedrooms, age in yrs, Distance to town(km)
neighbour_features= np.array([[2100, 3, 5, 2.5 ], [1400, 2, 25, 12.0], [1900, 3, 10, 4.8], [2700, 4, 2, 1.5]], dtype=float)
```

```python
# Users input
size= input('Enter size in squre meters: ')
bedrooms=input( 'Enter prefered number of bedrooms: ')
age= input('Prefered age: ')
distance_from_town= input(' Enter prefered distance from town: ')
```

```python
# Real market prices for the house, indexed the same
neighbour_prices= np.array([65000, 45000, 34000, 102000])
# The targrt home features
target_home= np.array([float(size), float(bedrooms), float(age), float(distance_from_town)])
```

```python
# Normalise my data(brings all scales between 0 and 1)
# This prevents the larger data from overpowering the smaller data
data_min= neighbour_features.min(axis=0)
data_max= neighbour_features.max(axis=0)
# To normalize:
norm_neighbours= (neighbour_features - data_min)/(data_max- data_min)
norm_target= (target_home- data_min) / (data_max- data_min)
```

```python
# Calculate similarity and value
difference= norm_neighbours- norm_target
squared_difference= difference**2
summed_squares= np.sum(squared_difference, axis=1)
distances=np.sqrt(summed_squares)
```

```python
# finding the closest match in distance
closest_house= np.argmin(distances)
value= neighbour_prices[closest_house]
```

```python
#visual output
print('TARGET HOME')
print('Match found: House#', closest_house)
print('Match Specs: ', neighbour_features[closest_house])
print('Match Prices: $',value)
print('Final Appraisal'        )
print('Estimated market value: ', value)
```
